```python
from misc import *
import constant
```

```python
# upto 5 long track candidates isMuon
LTRACK5_FILE_PATH = "/dice/projects/LHCb/PbPb_MSCi/isMuon_Tr1_5_B2JpsiK_PbPb/LeadLead_B2JpsiKTuple_all.root"
DEFAULT_TREE_NAME = "B2JpsiKTuple/DecayTree"

df = load_root(LTRACK5_FILE_PATH, DEFAULT_TREE_NAME)
df = df[df["Kplus_isMuon"] == False]
df = df[(constant.JPSI_MASS-50 <= df["J_psi_1S_M"]) & (df["J_psi_1S_M"] <= constant.JPSI_MASS+50)]
```

# Removing duplicates events where another candidate has a higher 3 body vertex quality

```python
best_envertex_df, _ = filter_duplicates_by_performance(
    df, "Bplus_ENDVERTEX_CHI2", "min"
)

histogram_fig(
    [
        first := HistogramPlot(df["Bplus_M"], "All"),
        HistogramPlot(
            best_envertex_df["Bplus_M"],
            r"Minimum $B^+$ ENDVERTEX $\chi^2$",
            bins=first.bin_edges,
        ),
    ],
    "figures/week14/duplicates_removed_endvertex_full.png",
    xlabel="Candidate Mass",
    #  xlim=[constant.BPLUS_MASS - 100, constant.BPLUS_MASS + 100],
    unit_tex=r"\mathrm{MeV}/c^2",
    #  ylim=[0,25]
    xlim=[3000, 8000],
)

histogram_fig(
    [
        first := HistogramPlot(
            df["Bplus_M"],
            "All",
            bins=5,
            error=True,
            alpha=0.4,
            linewidth=1.5,
        ),
        HistogramPlot(
            best_envertex_df["Bplus_M"],
            r"Minimum $B^+$ ENDVERTEX $\chi^2$",
            bins=first.bin_edges,
            error=True,
            linewidth=1.5,
        ),
    ],
    "figures/week14/duplicates_removed_endvertex_Bmass.png",
    preset="Bplus Mass",
    ylim=[0, 20],
)

histogram_fig(
    [
        first := HistogramPlot(
            df["Bplus_M"],
            "All",
            bins=15,
            error=True,
            alpha=0.4,
            linewidth=1.5,
        ),
        HistogramPlot(
            best_envertex_df["Bplus_M"],
            r"Minimum $B^+$ ENDVERTEX $\chi^2$",
            bins=first.bin_edges,
            error=True,
            linewidth=1.5,
        ),
    ],
    "figures/week14/duplicates_removed_endvertex_Bmass.png",
    preset="Bplus Mass",
    ylim=[0, 30],
)
```

# Removing duplicates events where another candidate has a higher PIDK

```python
best_pidk_df, _ = filter_duplicates_by_performance(df, "Kplus_PIDK", "max")

histogram_fig(
    [
        first := HistogramPlot(df["Bplus_M"], "All"),
        HistogramPlot(
            best_pidk_df["Bplus_M"], "Best PIDK quality candidate", bins=first.bin_edges
        ),
    ],
    "figures/week14/duplicates_removed_pidk_full.png",
    xlabel="Mass",
    #  xlim=[constant.BPLUS_MASS - 100, constant.BPLUS_MASS + 100],
    unit_tex=r"\mathrm{MeV}/c^2",
    #  ylim=[0,25]
    xlim=[3000, 7000]
)

histogram_fig(
    [
        first := HistogramPlot(
            df["Bplus_M"],
            "All",
            bins=5,
            error=True,
            alpha=0.4,
            linewidth=1.5,
        ),
        HistogramPlot(
            best_pidk_df["Bplus_M"],
            "Best PIDK quality candidate",
            bins=first.bin_edges,
            error=True,
            alpha=0.9,
            linewidth=1.5,
        ),
    ],
    "figures/week14/duplicates_removed_pidk_Bmass.png",
    #  xlabel="Mass",
    #  xlim=[constant.BPLUS_MASS - 100, constant.BPLUS_MASS + 100],
    #  unit_tex=r"\mathrm{MeV}/c^2",
    preset="Bplus Mass",
    ylim=[0, 21],
)

histogram_fig(
    [
        first := HistogramPlot(
            df["Bplus_M"],
            "All candidates",
            bins=5,
            error=True,
            alpha=0.4 ,
            linewidth=1.5,
        ),
        HistogramPlot(
            best_pidk_df["Bplus_M"],
            "Maximum PIDK candidate",
            bins=first.bin_edges,
            error=True,
            alpha=0.9,
            linewidth=1.5,
        ),
    ],
    False,
    preset="Bplus Mass",
    ylim=[0, 28],
)

histogram_fig(
    [
        first := HistogramPlot(
            df["Bplus_M"],
            "All",
            bins=15,
            error=True,
            alpha=0.4 ,
            linewidth=1.5,
        ),
        HistogramPlot(
            best_pidk_df["Bplus_M"],
            "Best PIDK quality candidate",
            bins=first.bin_edges,
            error=True,
            alpha=0.9,
            linewidth=1.5,
        ),
    ],
    False,
    preset="Bplus Mass",
    ylim=[0, 28],
)
```
